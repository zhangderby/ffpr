```python
platescale = 4.905 # arcsec/mm

w = 36 * platescale / 60
h = 24 * platescale / 60
w_offset = w / 2
h_offset = h / 2

positions = [(data['xy'][0] + w_offset, data['xy'][1] + h_offset) for data in detector_data]
defocuses = [float(data['defocus']) for data in detector_data]
```

```python
field_interps_nom, fields_x, fields_y = create_field_interps()

sys = cobalt(opd_maps=opd_maps, config_stp=cfg_um, config_wcc=cfg_um_wcc,
             data_path_stp=path_um, data_path_wcc=path_um_wcc, field_interps_nom=field_interps_nom)

# lots n lots o stars
mags = (np.ones(5) * 15).tolist()
positions = [(-0.32, 0.13), (-0.32, -0.13), (0, 0), (0.32, 0.13), (0.32, -0.13)]
defocuses = (np.zeros(5)).tolist()
sys.set_source_parameters(magnitudes=mags, positions=positions, defocus_vals=defocuses)

# sys.set_m1_bending(-np.sum(np.array(sys.get_opds()['Optics']), axis=0))
sys.set_cam_exposure_time(1)

images = sys.snap(stacked_frames=10)

imnorm = images[1].max()

print(imnorm)
```

```python
plt.figure(figsize=(20, 4))
count = 1
for im, pos in zip(images, positions):
    plt.subplot(1, 5, count)
    plt.imshow(im.get(), cmap='magma', norm='log', vmax=sys.well_depth, vmin=1)
    plt.colorbar(fraction=0.046, pad=0.04)
    plt.title(pos)
    count += 1
```

```python
opds = sys.get_opds()['Field']

optic_opd = np.sum(np.array(sys.get_opds()['Optics']), axis=0)

rms_vals = []
for opd in opds:
    opd[~sys.pupil] = np.nan
    opd -= np.nanmean(opd)
    rms_vals.append(np.sqrt(np.nanmean(opd ** 2)))

v1 = np.nanmax(np.array(opds))
v2 = np.abs(np.nanmin(np.array(opds)))
v = np.max(np.array([v1, v2]))

plt.figure(figsize=(20, 4))
count = 1
for opd, rms in zip(opds, rms_vals):
    plt.subplot(1, 5, count)
    plt.imshow(opd.get(), cmap='RdBu', vmin=-v, vmax=v)
    plt.colorbar(fraction=0.046, pad=0.04)
    plt.title(rms)
    count += 1
```

```python
optic_opd = np.sum(np.array(sys.get_opds()['Optics']), axis=0)

optic_opd[~sys.pupil] = np.nan

v1 = np.nanmax(optic_opd)
v2 = np.abs(np.nanmin(optic_opd))
v = np.max(np.array([v1, v2]))

plt.figure(figsize=(5, 5), dpi=150)
im = plt.imshow(optic_opd.get(), cmap='coolwarm', vmin=-v, vmax=v, extent=[-1.5, 1.5, -1.5, 1.5])
cb = plt.colorbar(im, fraction=0.046, pad=0.04)
cb.set_label('OPD (nm)', fontsize=14)
cb.ax.tick_params(labelsize=14)
plt.xlabel('Meters', fontsize=14)
plt.xticks(fontsize=14)
plt.yticks([])
plt.title(f'M1 - M4 Optic Surface Errors\n{np.sqrt(np.nanmean(optic_opd[sys.pupil] ** 2)):.2f} nm RMS', fontsize=16)
```

```python
# zernike basis (excluding piston) up to Z37
x, y = make_xy_grid(sys.pupil.shape[0], dx=sys.dx_pup)
r, t = cart_to_polar(x, y)
r_norm = r / (sys.dx_pup * sys.npix / 2)

nms = [noll_to_nm(i) for i in range(2, 11)]
zernikes = np.array(zernike_nm_seq(nms, r_norm, t, norm=True))
# zernikes = np.array([z / np.max(np.abs(z)) for z in zernikes])

inds = [i + 2 for i in range(8)]
```

# closed loop ffpr

```python
opd_rms = 10

coeffs_opd = np.random.normal(loc=0, size=len(zernikes[inds]))
input_opd = sum_of_2d_modes(zernikes[inds], coeffs_opd)
input_opd *= opd_rms / np.sqrt(np.mean(input_opd[sys.pupil] ** 2))
input_opd[~sys.pupil] = 0

plt.imshow(input_opd.get())
plt.colorbar()
plt.title(f'{np.sqrt(np.nanmean(input_opd[sys.pupil] ** 2))}')
```

```python
sys.set_m1_bending(input_opd) # - optic_opd)

images = sys.snap(stacked_frames=10)

plt.figure(figsize=(20, 4))
count = 1
for im, pos in zip(images, positions):
    plt.subplot(1, 5, count)
    plt.imshow(im.get(), cmap='magma', norm='log', vmax=sys.well_depth, vmin=1)
    plt.colorbar(fraction=0.046, pad=0.04)
    plt.title(pos)
    count += 1
```

```python
adpr_list = []
for psf, defocus in zip(images, defocuses):
    adpr_list.append(ADPR(wvls=sys.wvls.tolist(), pupil=sys.pupil, dx_pupil=sys.dx_pup, psf=psf, dx_psf=sys.dx_detector,
                        efl=sys.efl, modes=zernikes, defocus_coeff=defocus))

ff_opt = FFPR(optlist=adpr_list, psf_positions=sys.src_pos, field_modes=zernikes, field_coeff_interps=field_interps_nom)

ff_opt.minimize_common(jac=True, method='L-BFGS-B', options={'maxls' : 10, 'ftol' : 1e-20, 'gtol' : 1e-8, 'disp' : 0, 'maxiter' : 20})

plt.figure(figsize=(14, 8))
plt.title('FFPR Optimization', fontsize=16)
for i, opt in enumerate(ff_opt.optlist):
    plt.semilogy(np.array(opt.costs).get(), label=f'({sys.src_pos[i][0] * 60:.1f}", {sys.src_pos[i][1] * 60:.1f}")', alpha=0.5, linewidth=1)
plt.semilogy(np.array(ff_opt.costs).get(), label='Mean', color='k', linewidth=1)
plt.xlabel('L-BFGS-B Iteration', fontsize=14)
plt.ylabel('Error Function', fontsize=14)
plt.legend(ncol=3, fontsize=14)
plt.xlim([0, len(ff_opt.costs)])
plt.ylim([0, 1])
plt.grid(alpha=0.4, which='both')
plt.xticks(fontsize=14)
plt.yticks(fontsize=14)
```
